"https://accounts.spotify.com/api/token" with as parameters grant_type = "client_credentials", client_id
your Client ID and client_secret your Client Secret. Store the access token that you retrieve.
Note that this access token does not grant access to your personal account details, only to more general data
about artists and songs.
You can find more details here: https://developer.spotify.com/documentation/web-api/tutorials/getting-started
